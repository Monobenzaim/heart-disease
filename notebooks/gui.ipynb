```python
import tkinter as tk
from tkinter import messagebox
from mqtt_client import client, digital_twin

# GUI Setup
root = tk.Tk()
root.title("Heart Disease Digital Twin")
root.geometry("400x300")

# Display initial data
def show_data():
    current_state = digital_twin.get_current_state()
    data_text = "\n".join([f"{key}: {value}" for key, value in current_state.items()])
    data_label.config(text=data_text)

# Show prediction based on the current data
def show_prediction():
    current_state = list(digital_twin.get_current_state().values())
    prediction = client.model.predict([current_state])
    prediction_result = "Positive" if prediction[0] == 1 else "Negative"
    messagebox.showinfo("Prediction", f"Heart Disease Risk: {prediction_result}")

# Create the GUI layout
data_label = tk.Label(root, text="Patient Data:\n")
data_label.pack(pady=10)

refresh_button = tk.Button(root, text="Refresh Data", command=show_data)
refresh_button.pack(pady=5)

predict_button = tk.Button(root, text="Get Prediction", command=show_prediction)
predict_button.pack(pady=5)

# Start with the initial data
show_data()

# Start the GUI loop
root.mainloop()
```
