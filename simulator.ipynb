```python
import json
import time
import random
import paho.mqtt.client as mqtt

# MQTT settings
BROKER = "broker.hivemq.com"
PORT = 1883
TOPIC = "patient/data"

# Connect to MQTT broker
client = mqtt.Client()
client.connect(BROKER, PORT)

def generate_patient_data(patient_id):
    """
    Generate synthetic patient data.
    """
    return {
        "id": patient_id,
        "age": random.randint(30, 80),
        "sex": random.randint(0, 1),
        "cp": random.randint(0, 3),
        "trestbps": random.randint(90, 180),
        "chol": random.randint(150, 300),
        "fbs": random.randint(0, 1),
        "restecg": random.randint(0, 2),
        "thalach": random.randint(70, 200),
        "exang": random.randint(0, 1),
        "oldpeak": round(random.uniform(0.0, 6.0), 1),
        "slope": random.randint(0, 2),
        "ca": random.randint(0, 4),
        "thal": random.randint(1, 3)
    }

# Simulate patient data and publish to MQTT
while True:
    patient_id = random.randint(1000, 9999)
    patient_data = generate_patient_data(patient_id)
    client.publish(TOPIC, json.dumps(patient_data))
    print(f"Published data for patient {patient_id}")
    time.sleep(5)
```
